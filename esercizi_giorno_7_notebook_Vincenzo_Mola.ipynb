```python
import pandas as pd
import re
import numpy as np

df_dipendenti = pd.read_csv("dipendenti_sporco.csv")

df_dipendenti_copia = df_dipendenti.copy()

df_dipendenti_copia = df_dipendenti_copia.drop_duplicates()

def pulizia_stipendio(value):

    if pd.isna(value):
        return np.nan

    string_value = str(value).strip()
    caratteri_giusti = re.sub(r"[^0-9.,-]", "", string_value)
    caratteri_giusti = caratteri_giusti.replace(",", ".")

    try:
        numero = float(caratteri_giusti)
    except ValueError:
        return np.nan

    if "." in caratteri_giusti and len(caratteri_giusti.split(".")[1]) > 2:
        return np.nan

    return round(numero, 2)

df_dipendenti_copia['Nome'] = df_dipendenti_copia['Nome'].str.strip().str.lower().str.capitalize()
df_dipendenti_copia['Cognome'] = df_dipendenti_copia['Cognome'].str.strip().str.lower().str.capitalize()
df_dipendenti_copia['Email'] = df_dipendenti_copia['Email'].str.strip().str.lower()
df_dipendenti_copia['Data di Assunzione'] = pd.to_datetime(df_dipendenti_copia['Data di Assunzione'], format='mixed', errors='coerce')
df_dipendenti_copia["Stipendio"] = (df_dipendenti_copia["Stipendio"].apply(pulizia_stipendio).astype(float).round(2))
df_dipendenti_copia["id_dipartimento"] = pd.to_numeric(df_dipendenti_copia['id_dipartimento'], errors='coerce')

df_dipendenti_copia['Nome'] = df_dipendenti_copia['Nome'].fillna("Non Specificato")
df_dipendenti_copia['Cognome'] = df_dipendenti_copia['Cognome'].fillna("Non Specificato")
df_dipendenti_copia['Email'] = df_dipendenti_copia['Email'].fillna("Non Specificato")

print(df_dipendenti_copia)

display(df_dipendenti_copia)
df_dipendenti_copia.info()

```
